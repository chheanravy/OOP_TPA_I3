Exercise 1: Student Information System

```python
class Student:
    # Constructor
    def __init__(self, name, age, student_id):
        self.name = name
        self.age = age
        self.student_id = student_id

    # Instance method to display student information
    def details(self):
        print(f"Student Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Student ID: {self.student_id}")

    # Instance method to check whether the student is an adult
    def is_adult(self):
        if self.age >= 18:
            return "Yes"
        else:
            return "No"

# Create three student objects
student1 = Student("Visal", 22, "e2024")
student2 = Student("Dara", 17, "e2025")
student3 = Student("Sokha", 20, "e2026")

# Display information of each student
students = [student1, student2, student3]

for student in students:
    student.details()
    print(f"Adult: {student.is_adult()}")

"""Qestions and Answers"""
"""1. Why do we use a constructor?"""
"""We use it to initialize an object's attributes when it's created. Ensuring that each object has the necessary information"""
"""2. What is the role of instance methods?"""
"""Instance methods are functions defined inside a class that perform operations on indiidual objects."""
```

Exercise 2: Bank Account

```python
class BankAccount:

    # Class attribute
    bank_name = "ITC Bank"

    # Constructor
    def __init__(self, owner, account_number, balance):
        self.owner = owner
        self.account_number = account_number
        self.balance = balance

    # Instance method: Deposit money
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposit: ${amount}")
            print(f"New Balance: ${self.balance}")
        else:
            print("Error: Deposit amount must be positive.")

    # Instance method: Withdraw money
    def withdraw(self, amount):
        if amount <= 0:
            print("Error: Withdrawal amount must be positive.")
        elif amount > self.balance:
            print("Error: Insufficient balance.")
        else:
            self.balance -= amount
            print(f"Withdraw: ${amount}")
            print(f"Final Balance: ${self.balance}")

    # Instance method: Display account balance
    def display_balance(self):
        print(f"Bank: {BankAccount.bank_name}")
        print(f"Account Owner: {self.owner}")
        print(f"Account Number: {self.account_number}")
        print(f"Initial Balance: ${self.balance}")


# Create two bank account objects
account1 = BankAccount("Visal", "AC001", 1000)
account2 = BankAccount("Dara", "AC002", 2000)


# Test the first account
account1.display_balance()

print("\n--- Deposit Operation ---")
account1.deposit(500)

print("\n--- Withdrawal Operation ---")
account1.withdraw(200)

print("\n--- Second Account ---")
account2.display_balance()
```

Exercise 3: Rectangle Calculator

```python
# Practice 3: Rectangle Calculator
# Create class Rectangle
class Rectangle:
    # Constructor
    def __init__(self, width, length):
        self.width = width
        self.length = length

    # Instance method area()
    def area(self):
        return self.width * self.length

    # Instance method perimeter()
    def perimeter(self):
        return 2 * (self.width + self.length)

    # Create a static method
    @staticmethod
    def is_valid(width, length):
        return width > 0 and length > 0

# Check Rectangle 1
if Rectangle.is_valid(5, 10):
    rectangle1 = Rectangle(5, 10)
    print("Rectangle 1")
    print("Width:", rectangle1.width)
    print("Length:", rectangle1.length)
    print("Area:", rectangle1.area())
    print("Perimeter:", rectangle1.perimeter())

# Check Rectangle 2
if Rectangle.is_valid(4, 8):
    rectangle2 = Rectangle(4, 8)
    print("\nRectangle 2")
    print("Width:", rectangle2.width)
    print("Length:", rectangle2.length)
    print("Area:", rectangle2.area())
    print("Perimeter:", rectangle2.perimeter())

# Check Rectangle 3
if Rectangle.is_valid(6, 12):
    rectangle3 = Rectangle(6, 12)
    print("\nRectangle 3")
    print("Width:", rectangle3.width)
    print("Length:", rectangle3.length)
    print("Area:", rectangle3.area())
    print("Perimeter:", rectangle3.perimeter())

# Check Rectangle 4
if Rectangle.is_valid(-6, 12):
    rectangle4 = Rectangle(-6, 12)
    print("\nRectangle 3")
    print("Width:", rectangle4.width)
    print("Length:", rectangle4.length)
    print("Area:", rectangle4.area())
    print("Perimeter:", rectangle4.perimeter())

else:
    print("\nInvalid rectangle dimensions!")
```

Exercise 5: Temperature Converter

```python
class Temperature:

    # Static method: Celsius to Fahrenheit
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (9 / 5) * celsius + 32

    # Static method: Fahrenheit to Celsius
    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return (5 / 9) * (fahrenheit - 32)

    # Static method: Celsius to Kelvin
    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15


# Test the methods with different temperatures

celsius = 25

fahrenheit = Temperature.celsius_to_fahrenheit(celsius)
kelvin = Temperature.celsius_to_kelvin(celsius)

print(f"Celsius: {celsius}")
print(f"Fahrenheit: {fahrenheit}")
print(f"Kelvin: {kelvin}")

print("-" * 30)

fahrenheit_input = 77

celsius_result = Temperature.fahrenheit_to_celsius(fahrenheit_input)

print(f"Fahrenheit: {fahrenheit_input}")
print(f"Celsius: {celsius_result}")
```

Exercise 6: Product Management System

```python
class Product:

    # Class attributes
    store_name = "ITC Online Store"
    product_count = 0

    # Constructor
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

        # Increase product count whenever a new object is created
        Product.product_count += 1

    # Instance method: Display product information
    def display_product(self):
        print(f"Name: {self.name}")
        print(f"Price: ${self.price}")
        print(f"Quantity: {self.quantity}")
        print(f"Total Value: ${self.total_value()}")

    # Instance method: Calculate total product value
    def total_value(self):
        return self.price * self.quantity

    # Class method: Get total number of products
    @classmethod
    def get_product_count(cls):
        return cls.product_count

    # Static method: Validate product price
    @staticmethod
    def is_valid_price(price):
        return price > 0


# Create three Product objects
product1 = Product("Laptop", 800, 5)
product2 = Product("Mouse", 15, 20)
product3 = Product("Keyboard", 30, 10)


# Store all products in a list
products = [product1, product2, product3]


# Display store information
print("========== ITC ONLINE STORE ==========")
print()

# Calculate total inventory value
total_inventory_value = 0

for index, product in enumerate(products, start=1):

    print(f"Product {index}")

    # Check whether the product price is valid
    if Product.is_valid_price(product.price):
        product.display_product()
    else:
        print("Invalid product price!")

    total_inventory_value += product.total_value()

    print()


# Display summary
print(f"Total Products: {Product.get_product_count()}")
print(f"Total Inventory Value: ${total_inventory_value}")


"""Questions and Answers"""
"""1. What is the difference between instance attributes and class attributes?"""
"""Instance attributes are vairables that belong to individual objects. Each object maintians its own values,
so modifying one object's instance attribute does not directly affect other objects.
Class attributes are variables that belong to the class itself and are shared among its instances.
They are useful for storing comon information that multiple objects need to access."""
"""2. Why should we use a class method to count products?"""
"""We use class method to count products because the total number of products belongs to the entire class rather than
an individual object."""
"""3. Why is is_valid_price() implemented as a static method>"""
"""The is_valid_price() method is implementeed as a static mehtod because checking whether a price is positve deos not 
require acceess to instance attributes or class attributes. It only needs the price provided as an argument."""
"""4. Which methods in this exercisee use self, cls, or neither?"""
"""There are 3 types of methods: __init__, display_product(), total_value(), get_product_count() and is_valid_price()"""
```
